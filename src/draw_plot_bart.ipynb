```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('../data/reverse_experiments/june_version_7921032488/results/bart_sim_acc_summary_base.csv')

# 过滤掉包含 'reverse' 的 key 名称
reverse_df = df[df['key'].str.contains('reverse')]
normal_df = df[~df['key'].str.contains('reverse')]

# 设置柱状图的画布大小
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# 第一张图：acc_bart
ax[0].bar(normal_df['key'], normal_df['acc_bart'], label='Normal', color='skyblue')
ax[0].bar(reverse_df['key'], reverse_df['acc_bart'], label='Reverse', color='salmon')

ax[0].set_title('Accuracy (acc_bart)')
ax[0].set_xlabel('Test Sets')
ax[0].set_ylabel('Accuracy')
ax[0].tick_params(axis='x', rotation=45)
ax[0].legend()

# 第二张图：sim_acc_bart
ax[1].bar(normal_df['key'], normal_df['sim_acc_bart'], label='Normal', color='skyblue')
ax[1].bar(reverse_df['key'], reverse_df['sim_acc_bart'], label='Reverse', color='salmon')

ax[1].set_title('Similarity Accuracy (sim_acc_bart)')
ax[1].set_xlabel('Test Sets')
ax[1].set_ylabel('Similarity Accuracy')
ax[1].tick_params(axis='x', rotation=45)
ax[1].legend()

# 自动调整布局
plt.tight_layout()

# 显示图表
plt.show()
```
